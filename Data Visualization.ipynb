# **Data Visualization using Python**

```python
# Basic plots
# importing required modules
%matplotlib inline
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 50) # create 50 equi-distant numbers between 0 to 10 (inclusive)
print(x)
```

```python
x = np.linspace(0, 10, 50)
sinus = np.sin(x)
plt.plot(x, sinus)
plt.show()
```

```python
plt.plot(x, sinus, "o")
plt.show()
```

```python
plt.plot(x, sinus, "--")
plt.show()
```

```python
plt.plot(x, sinus, "*")
plt.show()
```

```python
plt.plot(x, sinus, "-")
plt.show()
```

```python
help(plt.plot)
```

```python
# rapid multi-plot
x = np.linspace(0, 10, 50)
sinus = np.sin(x)
cosinus = np.cos(x)
plt.plot(x, sinus, "-b", x, sinus, "or", x, cosinus, "--r", x, cosinus, "ob")
plt.xlabel("X-Values...")
plt.ylabel("Sin(x) / Cos(x) values...")
plt.title("Plotting Sin(x) and Cos(x) vs. x-values...")
plt.show()
```

```python
# step by step plotting
x = np.linspace(0, 10, 50)
sinus = np.sin(x)
cosinus = np.cos(x)
plt.plot(x, sinus, label="sinus", color="blue", linestyle="--", linewidth=2)
plt.plot(x, cosinus, label="cosinus", color="red", linestyle="-", linewidth=2)
plt.legend(loc=0)
plt.show()
```

```python
# Class Assignment:
# Draw a circle with red color with radius = 10 and center at the origin (0, 0).
# x^2 + y^2 = r^2
```

```python
r = 10
theta = np.linspace(0, 2 * np.pi, 100)
x = r * np.cos(theta)
y = r * np.sin(theta)
plt.plot(x, y, color="red")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Circle with radius 10")
plt.axis("equal")
plt.show()
```

```python
# x^2 + y^2 = r^2
r = 10
x = np.linspace(-10, 10, 100)
y = np.sqrt(r**2 - x**2)
plt.figure(figsize=(4, 4))
plt.plot(x, y, "-r")
plt.plot(x, -y, "-r")
# plt.axis("equal")
plt.show()
```

## **Scatter 2D Plot**

```python
try:
    # salary_df = pd.read_csv("salary_table.csv")
    salary_df = pd.read_csv("https://raw.githubusercontent.com/toarnabtrainer/SVGI-Data-Science-Course-Batch-1/refs/heads/main/Datafile/salary_table.csv")
    print("Dimension is:", salary_df.shape)
    display(salary_df.head())
except:
    print("File access error...")
```

```python
print(salary_df.education.value_counts())
colors_edu = {"Bachelor":"r", "Master":"g", "Ph.D":"b"}
plt.scatter(salary_df.education, salary_df.salary, c = salary_df.education.apply(lambda edu: colors_edu[edu]), s = 100)
plt.show()
```

```python
colors_edu = {"Bachelor":"r", "Master":"g", "Ph.D":"b"}
plt.scatter(salary_df["experience"], salary_df["salary"], c = salary_df.education.apply(lambda edu: colors_edu[edu]), s = 100)
plt.show()
```

## **Scatter plot with colors and symbols**

```python
symbol_manage = {"Y":"*", "N":"."}
colors_edu = {"Bachelor":"r", "Master":"g", "Ph.D":"b"}
print(salary_df.columns)
print(salary_df.management.value_counts())
print(salary_df.education.value_counts())
```

```python
plt.figure(figsize=(6, 5))
# group by education x management => 6 groups
for values, data in salary_df.groupby(['education', 'management']):
    print(values)
    display(data)
```

```python
plt.figure(figsize=(6, 5))
# group by education x management => 6 groups
for values, data in salary_df.groupby(['education', 'management']):
    edu, manager = values
    plt.scatter(data['experience'], data['salary'], marker=symbol_manage[manager], s=150, color=colors_edu[edu], label=manager+" / "+edu)
plt.legend()
plt.show()
```
